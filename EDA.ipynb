```python
# Replace 'male_players.csv' with whatever the main file name is in your list
csv_file_path = os.path.join(base_path, "players_22.csv") 
df = pd.read_csv(csv_file_path)

print(f"Dataset shape: {df.shape[0]} rows, {df.shape[1]} columns.")
```

```python
# 1. Check data types and overall non-null counts
print("--- DATASET INFO ---")
df.info()

# 2. Check summary statistics for numerical columns
print("\n--- SUMMARY STATISTICS ---")
display(df.describe().T[['count', 'mean', 'std', 'min', 'max']])

# 3. Find columns with missing values and calculate percentages
missing_data = df.isnull().sum()
missing_percentage = (missing_data / len(df)) * 100

missing_df = pd.DataFrame({'Missing Values': missing_data, 'Percentage (%)': missing_percentage})
missing_df = missing_df[missing_df['Missing Values'] > 0].sort_values(by='Percentage (%)', ascending=False)

print(f"\n--- COLUMNS WITH MISSING VALUES ({len(missing_df)} columns found) ---")
display(missing_df.head(15))
```

```python
# 1. Create a clean copy of the dataframe to work on
df_clean = df.copy()

# 2. Fill missing numerical performance stats with 0 (since they belong to players who don't record them, like GKs)
num_cols = df_clean.select_dtypes(include=['float64', 'int64']).columns
df_clean[num_cols] = df_clean[num_cols].fillna(0)

# 3. Fill missing categorical descriptions with 'None' or 'Unknown'
cat_cols = df_clean.select_dtypes(include=['object']).columns
df_clean[cat_cols] = df_clean[cat_cols].fillna('Unknown')

print(f"Missing values remaining: {df_clean.isnull().sum().sum()}")
```

```python
# Let's check outliers in 'value_eur' using IQR method
Q1 = df_clean['value_eur'].quantile(0.25)
Q3 = df_clean['value_eur'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Count outliers
outliers = df_clean[(df_clean['value_eur'] < lower_bound) | (df_clean['value_eur'] > upper_bound)]
print(f"Identified {len(outliers)} outliers in player market values using IQR (Bound > {upper_bound:,.0f} EUR).")

# Mathematical justification: Instead of dropping elite players (which deletes valuable domain data), 
# we cap/winsorize them at the upper bound so they don't skew machine learning models later.
df_clean['value_eur_capped'] = df_clean['value_eur'].clip(upper=upper_bound)
print("Outliers successfully treated via Capping (Winsorization)!")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 8))

# Select key numerical features for correlation
key_features = ['age', 'potential', 'wage_eur', 'value_eur', 'shooting', 'passing', 'defending', 'physic']
corr_matrix = df_clean[key_features].corr()

# Create Heatmap
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', linewidths=0.5, cbar_kws={'label': 'Correlation Coefficient'})
plt.title('Correlation Heatmap of FIFA 22 Player Attributes', fontsize=14, pad=15)
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig('correlation_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
```

Observation: There is a strong positive correlation ($0.82$) between player potential and market value_eur. Interestingly, wage_eur also scales heavily with market value.

Practical Implication: Clubs evaluating young prospects should prioritize high-potential tracking metrics, as market evaluation is directly tethered to projected player growth.

```python
plt.figure(figsize=(10, 5))
sns.histplot(df_clean['age'], bins=25, kde=True, color='teal')
plt.title('Distribution of Player Ages', fontsize=14, pad=15)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Player Count', fontsize=12)
plt.tight_layout()
plt.savefig('age_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

Observation: The age distribution is right-skewed, with the vast majority of professional players concentrated between 21 and 28 years old.

Practical Implication: Squad planning should account for a talent drop-off or retirement cluster as players cross the 30-year threshold.

```python
plt.figure(figsize=(12, 6))
# Filter top common positions for clarity
top_positions = ['ST', 'CB', 'CM', 'CAM', 'GK', 'CDM', 'LB', 'RB']
subset_df = df_clean[df_clean['team_position'].isin(top_positions)]

sns.boxplot(x='team_position', y='overall', data=subset_df, palette='Set2')
plt.title('Overall Ratings by Player Position', fontsize=14, pad=15)
plt.xlabel('Team Position', fontsize=12)
plt.ylabel('Overall Rating', fontsize=12)
plt.tight_layout()
plt.savefig('position_boxplots.png', dpi=300, bbox_inches='tight')
plt.show()
```
